# Data Structures and Algorithms

### Choosing a Data Structure

A data structure is a choice about **how data is organized so certain operations become efficient**. There is no best structure for every problem; choose based on the operations used most often.

Evaluate a data structure by:
- **Organization:** how elements are arranged — sequentially, hierarchically, or as a network.
- **Operations:** insert, delete, search, traverse, update — and their time complexity.
- **Constraints:** rules on how data can be accessed or modified.

**Key idea:** choose the structure that makes your most important operations efficient.

### Data Structure Categories

<details>
<summary>Linear vs. Nonlinear</summary>

**Linear structures** arrange elements sequentially.  
Each element typically connects to the next/previous element.

- **Array:** fast index access
- **Linked List:** flexible insert/delete
- **Stack:** LIFO
- **Queue:** FIFO

**Nonlinear structures** organize elements hierarchically or as networks.

- **Tree:** parent-child hierarchy
- **BST:** ordered tree for efficient search
- **Heap:** priority-based tree; min/max at root
- **Graph:** arbitrary relationships between nodes

</details>

<details>
<summary>Hash-Based Structures</summary>

A **hash table** maps `key → value` using a hash function.

- Lookup: `O(1)` average
- Insert: `O(1)` average
- Delete: `O(1)` average
- Worst case: `O(n)` due to collisions

**Think:** fast lookup by key → hash table / `dict`

</details>

<details>
<summary>Static vs. Dynamic</summary>

- **Static:** fixed size after creation → predictable and memory-efficient.
- **Dynamic:** can grow/shrink at runtime → more flexible, some extra overhead.

Python's `list` is a **dynamic array**: it automatically increases its capacity when needed.

</details>

### Major Categories of Algorithms

Algorithms can be grouped by their **problem-solving strategy**. Recognizing the pattern of a problem helps narrow down which algorithms to consider.

| Category | Core Idea | Common Examples |
|---|---|---|
| **Searching** | Find an element or determine whether it exists | Linear Search, Binary Search |
| **Sorting** | Arrange elements in a defined order | Bubble, Selection, Insertion, Merge, Quick, Heap Sort |
| **Recursion** | Solve a problem using smaller versions of itself | Tree traversal, Fibonacci, combinations |
| **Divide & Conquer** | Split → solve parts → combine results | Merge Sort, Quick Sort, Binary Search |
| **Greedy** | Make the best local choice at each step | Dijkstra*, Huffman Coding |
| **Graph Traversal** | Systematically visit connected nodes | BFS, DFS |

\* Dijkstra's algorithm is greedy when edge weights are non-negative.

#### Pattern Recognition

The important skill is not memorizing algorithms — it is recognizing **what kind of problem you are looking at**.

- **Find an item** → Searching
- **Put items in order** → Sorting
- **Same problem on smaller inputs** → Recursion
- **Split into independent parts** → Divide & Conquer
- **Best choice at each step** → Greedy
- **Nodes + connections / paths** → Graph algorithms

**Key idea:**  
`Recognize the problem pattern → choose the algorithm family → choose the specific algorithm.`

Before writing code, it helps to write out the logic in plain language!

### 1) Linked List

<details>

<summary>A chain of nodes: each node knows where the next one is</summary>

**Idea:**
A chain of **nodes**. Each node stores a value and a reference to the next node.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Frequent insertions/deletions
* Following `next` pointers
* Reversing a list
* Detecting a cycle
* Finding the middle

</details>

**Python:**
Python has no built-in linked list, so it is usually represented using a custom node:

`Node(value, next)`

**Complexity:**
<details>
<summary>Show the answer</summary>

* Access/Search: `O(n)`
* Insert/Delete when the location is known: `O(1)`
* If the location must first be found: `O(n)`

</details>

**Important:**
Unlike a Python list, there is no `O(1)` random access. To reach a node, start at the head and follow `next`.

</details>

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList():
    def __init__(self):
        self.head = None
        self.tail = None

    def insert_at_beginning(self, data):
        new_node = Node(data)
        if self.head:
            new_node.next = self.head
            self.head = new_node
        else:
            self.head = new_node
            self.tail = new_node

    def remove_at_beginning(self):
        self.head = self.head.next
```

**Q1) Middle Node:**&nbsp;&nbsp;Given the head of a singly linked list, return the middle node.

<details>
<summary>Show answer</summary>

Use two pointers:
- `slow` moves 1 step
- `fast` moves 2 steps
- When `fast` reaches the end, `slow` is at the middle

</details>

```python
def find_middle(head):
    slow = head
    fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

    return slow
```

### 2) Stack

<details>
<summary>LIFO (last in, first out)</summary>

**Idea:**
A singly linked list of elements with LIFO. Elements are added and removed from the top. The most recently added element is the first one removed (like a stack of books).

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Need the most recent item first
* Undo/history operations
* Matching or nested structures
* DFS / backtracking
* Reversing or processing things in reverse order

</details>

**Python:**
DataCamp uses Python's LifoQueue:

`queue.LifoQueue()`

A regular Python list can also be used as a stack with append() and pop().

**Complexity:**
<details>
<summary>Show the answer</summary>

* Access/Search: `O(n)`
* Push: `O(1)`
* Pop: `O(1)`
* Peek(top): `O(1)`

</details>

**Important:**
A stack only gives direct access to the top element. To reach an item deeper in the stack, elements above it must be processed first.

</details>

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class Stack:
    def __init__(self):
        self.top = None
        self.size = 0

    def push(self, data):
        new_node = Node(data)
        if self.top:
            new_node.next = self.top
        self.top = new_node
        self.size += 1   

    def pop(self):
        if self.top is None:
            return None
        else:
            popped_node = self.top
            self.size -=1
            self.top = self.top.next
            popped_node.next = None
            return popped_node.data

    def peek(self):
        if self.top:
            return self.top.data
        else:
            return None
```

```python
# use Python's built-in LifoQueue class to implement a stack
import queue

# create an infinite stack using LifoQueue
stack = queue.LifoQueue(maxsize=0)

# add elements to the stack
stack.put("first")
stack.put("second")

# get element
print("Top element:", stack.get())

# show the size
print("Size of stack:", stack.qsize())

# check if empty
print("Empty stack: ", stack.empty())
```

### 3) Queue

<details>
<summary>FIFO (first in, first out)</summary>

**Idea:**
Elements are added at the back and removed from the front. The first element added is the first one removed.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Items must be processed in arrival order
* Tasks/jobs waiting to be handled
* BFS / level-by-level traversal
* Scheduling or buffering

</details>

**Python:**
DataCamp uses Python's SimpleQueue:

`queue.SimpleQueue()`

A `collections.deque` is also commonly used as a queue.

**Complexity:**
<details>
<summary>Show the answer</summary>

* Enqueue: O(1)
* Dequeue: O(1)
* Access front: O(1)
* Search: O(n)

</details>

**Important:**
A queue follows FIFO: elements leave in the same order they entered.

</details>

```python
class Node():
    def __init__(self, data):
        self.data = data
        self.next = None

class Queue():
    def __init__():
        self.head = None
        self.tail = None
        
    def enqueue(self, data):
        new_node = Node(data)
        if self.head == None:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            self.tail = new_node 
    
    def dequeue(self):
        if self.head:
            current_node = self.head
            self.head = current_node.next
            current_node.next = None
        else:
            self.tail = None
```

```python
# use Python's built-in SimpleQueue class to implement a queue
import queue

# create an orders queue using SimpleQueue
orders_queue = queue.SimpleQueue()

# add elements to the queue
orders_queue.put("first")
orders_queue.put("second")
orders_queue.put("third")

# get element
print("Top element:", orders_queue.get())

# show the size
print("Size of queue:", orders_queue.qsize())

# check if empty
print("Empty queue: ", orders_queue.empty())
```

### 5) Tree

<details>
<summary>A hierarchical graph with no cycles</summary>

**Idea:**
A tree is a connected graph with no cycles. Nodes have parent-child relationships, starting from a root.
A binary tree allows each node to have at most two children.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Data has a hierarchical structure (like html)
* Parent-child relationships
* Searching/traversing levels or branches
* Problems involving paths, depth, or subtrees (file structure in computer)

</details>

**Python:**
Usually represented with a custom node class.

`Node(value, left, right)` for a binary tree.

**Complexity:**
<details>
<summary>Show the answer</summary>

* Traverse all nodes: O(n)
* Search in a general binary tree: O(n)
* Search in a balanced Binary Search Tree: O(log n) average

</details>

**Important:**
A binary tree is not automatically a Binary Search Tree (BST).
A BST additionally follows an ordering rule between the left child, node, and right child.

</details>

```python
class TreeNode:
    def __init__(self, data, left=None, right=None):
        self.data = data
        self.left_child = left
        self.right_child = right
    
    def is_leaf(self):
        return self.left_child is None and self.right_child is None

    def print_tree(self, level=0):
        if self.right_child:
            self.right_child.print_tree(level + 1)
        print(' ' * 4 * level + '->', self.data)
        if self.left_child:
            self.left_child.print_tree(level + 1)

node1 = TreeNode("B")
node2 = TreeNode("C")
root_node = TreeNode("A", node1, node2)
root_node.print_tree()
```

### 6) Graph

<details>
<summary>Nodes connected by edges</summary>

**Idea:**
A graph consists of nodes (vertices) connected by edges. Unlike trees, graphs may contain cycles and disconnected nodes. Graphs can be:
- Directed or undirected
- Weighted or unweighted
- Connected or disconnected

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Representing relationships or connections
* Networks, roads, social connections, dependencies
* Finding paths between nodes
* Exploring connected components
* Problems involving BFS or DFS

</details>

**Python:**
Often represented using an adjacency list, commonly with a dictionary:

`{node: [neighbors]}`

**Complexity:**
<details>
<summary>Show the answer</summary>

* BFS means Breadth-First Search. It explores a graph level by level, visiting all nearby nodes before moving farther away.
* DFS means Depth-First Search. It follows one path as far as possible before backtracking and trying another path.
* Both BFS and DFS usually visit every node and edge once, so their time complexity is O(V + E).
* Space means the extra memory needed while storing the graph and performing the traversal. With an adjacency list, the graph itself uses O(V + E) space.

V = number of vertices, or nodes
E = number of edges, or connection

</details>

**Important:**
A graph does not have to connect every node, and edges may have a direction or a weight depending on the problem.

</details>

```python
class Graph:
    def __init__(self):
        self.vertices = {}
    
    def add_vertex(self, vertex):
        if vertex not in self.vertices:
            self.vertices[vertex] = []
    
    def add_edge(self, source, target):
        if source in self.vertices and target in self.vertices:
            self.vertices[source].append(target)
            self.vertices[target].append(source)  # For undirected graph

    def add_directed_edge(self, source, target):
        if source in self.vertices and target in self.vertices:
            self.vertices[source].append(target)
    
    def add_weighted_edge(self, source, target, weight):
        if source in self.vertices and target in self.vertices:
            self.vertices[source].append((target, weight))
            self.vertices[target].append((source, weight))  # For undirected graph

my_graph = Graph()
my_graph.add_vertex('David')
my_graph.add_vertex('Miriam')
my_graph.add_vertex('Martin')
my_graph.add_edge('David', 'Miriam')
my_graph.add_edge('David', 'Martin')
my_graph.add_edge('Miriam', 'Martin')

print(my_graph.vertices)
```

### 7) Dynamic Programming

<details>
<summary>Save repeated work instead of calculating it again</summary>

**Idea:**
Break a problem into smaller overlapping subproblems and store their results so they are not recomputed.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* The same subproblems are solved repeatedly
* A recursive solution does lots of duplicate work
* A problem asks for an optimal value, count, or number of ways

</details>

**Common approaches:**

* Memoization (top-down): recursion + cache
* Tabulation (bottom-up): build results iteratively

**Important:**
DP often trades extra memory for much faster execution.

</details>

```python
cache = [None]*(100)

def fibonacci(n): 
    if n <= 1:
        return n
    
    # Check if the value exists
    if cache[n] is None:
        # Save the result in cache
        cache[n] = fibonacci(n-1) + fibonacci(n-2)
    return cache[n]

print(fibonacci(6))
```

```python
def hanoi(num_disks, from_rod, to_rod, aux_rod):
  # the base case
  if num_disks >= 1:
    # the calls to the hanoi function
    hanoi(num_disks - 1, from_rod, aux_rod, to_rod)
    print("Moving disk", num_disks, "from rod", from_rod,"to rod",to_rod)
    hanoi(num_disks - 1, aux_rod, to_rod, from_rod)   

num_disks = 4
source_rod = 'A'
auxiliar_rod = 'B'
target_rod = 'C'

hanoi(num_disks, source_rod, target_rod, auxiliar_rod)
```

### 8) Search Algorithms

#### Linear Search

<details>
<summary>Check elements one by one</summary>

**Idea:**
Go through the elements sequentially until the target is found.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Data is unsorted
* The collection is small
* No faster search structure is available

</details>

**Complexity:**
<details>

<summary>Show the answer</summary>

* Best case: O(1)
* Worst/Average case: O(n)

</details>

**Important:**
Works on both sorted and unsorted data, but may need to inspect every element.

</details>

```python
def linear_search(arr, target):
    for i, value in enumerate(arr):
        if value == target:
            return i
    return -1

print(linear_search([1,5,8,9,15,20,70,72], 5))
```

#### Binary Search

<details>
<summary>Repeatedly cut the search space in half</summary>

**Idea:**
Compare the target with the middle element and discard half of the remaining search space each step.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Data is sorted
* Fast searching is needed
* The search space can be repeatedly divided in half

</details>

**Complexity:**
<details>

<summary>Show the answer</summary>

* Best case: O(1)
* Worst/Average case: O(log n)

</details>

**Important:**
Binary search requires the data/search space to be ordered in a way that lets you safely eliminate half each step.

</details>

```python
def binary_search(ordered_list, target):
    first = 0
    last  = len(ordered_list) - 1
    
    while first <= last:
        middle = (first + last) // 2
        # check whether the search value equals the value in the middle
        if target == ordered_list[middle]:
            return True

        # check whether the search value is smaller than the value in the middle
        elif target < ordered_list[middle]:
            # set last to the value of middle minus one
            last = middle - 1

        else:
            first = middle + 1
    return False

print(binary_search([1,5,8,9,15,20,70,72], 5))
```

```python
def binary_search_recursive(ordered_list, target):
    # Define the base case
    if len(ordered_list) == 0:
        return False

    else:
        middle = len(ordered_list) // 2
        # check whether the search value equals the value in the middle
        if target == ordered_list[middle]:
            return True

        elif target < ordered_list[middle]:
            # call recursively with the left half of the list
            return binary_search_recursive(ordered_list[:middle], target)

        else:
            # call recursively with the right half of the list
            return binary_search_recursive(ordered_list[middle+1:], target)

print(binary_search_recursive([1,5,8,9,15,20,70,72], 5))
```

#### Binary Search Tree (BST)

<details>
<summary>Binary tree where left < node < right</summary>

**Idea:**
A Binary Search Tree is a binary tree with an ordering rule:

* Values in the left subtree are smaller
* Values in the right subtree are larger
This makes searching faster than in a general binary tree when the tree is reasonably balanced.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Need ordered data with search/insert/delete
* Need to repeatedly find smaller/larger values
* Tree structure should preserve sorted order

</details>

**Python:**

Usually implemented with a custom node:

`Node(value, left, right)`

**Complexity:**
<details>

<summary>Show the answer</summary>

If balanced:

* Search: O(log n)
* Insert: O(log n)
* Delete: O(log n)

Worst case if highly unbalanced:

* Search/Insert/Delete: O(n)

</details>

**Important:**
A binary tree is not automatically a BST. The BST ordering rule is what makes efficient searching possible.

</details>

```python
class TreeNode:
    def __init__(self, data, left=None, right=None):
        self.data = data
        self.left_child = left
        self.right_child = right

class BinarySearchTree:
    def __init__(self):
        self.root = None

    def search(self, target):
        current = self.root
        while current:
            if target == current.data:
                return True
            elif target < current.data:
                current = current.left_child
            else:
                current = current.right_child
        return False

    def insert(self, data):
        new_node = TreeNode(data)

        if self.root is None:
            self.root = new_node
            return True
            
        current = self.root
        
        while True:
            if data == current.data:
                return False # don't insert a duplicate

            elif data < current.data:
                if current.left_child is None:
                    current.left_child = new_node
                    return True
                else:
                    current = current.left_child
                
            else:
                if current.right_child is None:
                    current.right_child = new_node
                    return True
                else:
                    current = current.right_child

    def delete(self, target):
        if self.root is None:
            return False

        # case: target is the root
        if self.root.data == target:
            # no children
            if self.root.left_child is None and self.root.right_child is None:
                self.root = None

            # only right child
            elif self.root.left_child is None:
                self.root = self.root.right_child

            # only left child
            elif self.root.right_child is None:
                self.root = self.root.left_child

            # two children
            else:
                successor = self.root.right_child
                succ_parent = self.root

                while successor.left_child:
                    succ_parent = successor
                    successor = successor.left_child
                
                self.root.data = successor.data
                if succ_parent == self.root:
                    succ_parent.right_child = successor.right_child
                else:
                    succ_parent.left_child = successor.right_child

            return True
        
        parent = self.root

        while True:
            # decide direction to search
            if target < parent.data:
                current = parent.left_child
                is_left = True

            else:
                current = parent.right_child
                is_left = False

            # target does not exist
            if current is None:
                return False

            # target found
            elif target == current.data:

                # case 1: no children
                if current.left_child is None and current.right_child is None:
                    if is_left:
                        parent.left_child = None
                    else:
                        parent.right_child = None

                # case 2: only right child
                elif current.left_child is None:
                    if is_left:
                        parent.left_child = current.right_child
                    else:
                        parent.right_child = current.right_child

                # case 3: only left child
                elif current.right_child is None:
                    if is_left:
                        parent.left_child = current.left_child
                    else:
                        parent.right_child = current.left_child

                # case 4: two children
                else:
                    successor = current.right_child
                    succ_parent = current

                    while successor.left_child:
                        succ_parent = successor
                        successor = successor.left_child
                    
                    current.data = successor.data
                    if succ_parent == current:
                        succ_parent.right_child = successor.right_child
                    else:
                        succ_parent.left_child = successor.right_child

                return True

            else:
                parent = current

    def find_min(self):
        if self.root is None:
            return None

        current = self.root
        while current.left_child:
            current = current.left_child
        return current.data

    def find_max(self):
        if self.root is None:
            return None

        current = self.root
        while current.right_child:
            current = current.right_child
        return current.data

# Create BST
bst = BinarySearchTree()

# Insert values
values = [50, 30, 70, 20, 40, 60, 80, 65]

for value in values:
    bst.insert(value)

# Search
print("Search 40:", bst.search(40))   # True
print("Search 100:", bst.search(100)) # False

# Min / Max
print("Minimum:", bst.find_min())     # 20
print("Maximum:", bst.find_max())     # 80

# Duplicate
print("Insert duplicate 40:", bst.insert(40))  # False

# Delete a leaf node
print("Delete 20:", bst.delete(20))   # True
print("Search 20:", bst.search(20))   # False

# Delete a node with one child
print("Delete 60:", bst.delete(60))   # True
print("Search 60:", bst.search(60))   # False
print("Search 65:", bst.search(65))   # True

# Delete a node with two children
print("Delete 70:", bst.delete(70))   # True
print("Search 70:", bst.search(70))   # False

# Delete the root
print("Delete 50:", bst.delete(50))   # True
print("Search 50:", bst.search(50))   # False

# Delete a value that does not exist
print("Delete 100:", bst.delete(100)) # False

# Check remaining range
print("Minimum:", bst.find_min())
print("Maximum:", bst.find_max())
```

#### Depth-First Search (DFS)

<details>
<summary>Go as deep as possible before backtracking</summary>

**Idea:**
DFS explores one branch as far as possible, then backtracks and continues with another branch. Usually implemented with recursion or a stack.

**DFS on Trees:**
<details>

<summary>Show traversal types</summary>

For a node with left and right children:

* Pre-order: Node → Left → Right
* In-order: Left → Node → Right
* Post-order: Left → Right → Node

**When to use:**

* In-order: BST → values in ascending order
* Pre-order: Copy/serialize a tree, prefix expressions
* Post-order: Delete/process children before parent, postfix expressions

**Complexity:**

All three visit every node:

* Time: O(n)
* Space: O(h) recursion/stack

n = number of nodes
h = tree height

</details>

**DFS on Graphs:**
<details>

<summary>Show the answer</summary>

**Think of it when:**

* Exploring connected components
* Checking whether a path exists
* Exploring all reachable nodes
* Detecting cycles
* Backtracking-style problems

**Complexity:**

* Time: O(V + E)
* Space: O(V)

V = vertices
E = edges

</details>

**Important:**

For graphs, track visited nodes so cycles do not cause repeated/infinite traversal.
For trees, a visited set is usually unnecessary because each child is reached through one parent.

</details>

```python
class TraversableBinarySearchTree(BinarySearchTree):
    def in_order(self, current_node):
        if current_node:
            self.in_order(current_node.left_child)
            print(current_node.data, end=" ")
            self.in_order(current_node.right_child)

    def pre_order(self, current_node):
        if current_node:
            print(current_node.data, end=" ")
            self.pre_order(current_node.left_child)
            self.pre_order(current_node.right_child)

    def post_order(self, current_node):
        if current_node:
            self.post_order(current_node.left_child)
            self.post_order(current_node.right_child)
            print(current_node.data, end=" ")
  
bst = TraversableBinarySearchTree()
values = [50, 30, 70, 20, 40, 60, 80]
for value in values:
    bst.insert(value)

print("In-order:")
bst.in_order(bst.root)

print("\nPre-order:")
bst.pre_order(bst.root)

print("\nPost-order:")
bst.post_order(bst.root)
```

```python
def dfs(visited_vertices, graph, current_vertex):
    # check if current_vertex hasn't been visited yet
    if current_vertex not in visited_vertices:
        print(current_vertex, end=" ")
        # add current_vertex to visited_vertices
        visited_vertices.add(current_vertex)

        for adjacent_vertex in graph[current_vertex]:
            # call recursively with the appropriate values
            dfs(visited_vertices, graph, adjacent_vertex)
            
graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": [],
    "F": []
}

dfs(set(), graph, "A")
```

#### Breadth-First Search (BFS)

<details>
<summary>Explore level by level using a queue</summary>

**Idea:**
BFS explores all nearby nodes first before moving deeper. Usually implemented with a queue.

**BFS on Trees:**
<details>

<summary>Show the answer</summary>

**Think of it when:**

* Traversing a tree level by level
* Finding nodes by depth/level
* Finding the shortest path in an unweighted tree

**Complexity:**

* Time: O(n)
* Space: O(n)

n = number of nodes

</details>

**BFS on Graphs:**
<details>

<summary>Show the answer</summary>

**Think of it when:**

* Exploring nodes level by level
* Finding the shortest path in an unweighted graph
* Searching nearby nodes before going deeper
* Exploring connected components

**Complexity:**

* Time: O(V + E)
* Space: O(V)

V = vertices
E = edges

</details>

**Important:**

BFS uses a queue (FIFO).
For graphs, keep track of visited nodes to avoid revisiting nodes or getting stuck in cycles.

</details>

```python
import queue

class SearchableBinarySearchTree(TraversableBinarySearchTree):

    def bfs(self):
        if self.root is None:
            return []

        visited_nodes = []
        bfs_queue = queue.SimpleQueue()

        bfs_queue.put(self.root)

        while not bfs_queue.empty():
            current_node = bfs_queue.get()

            visited_nodes.append(current_node.data)

            if current_node.left_child:
                bfs_queue.put(current_node.left_child)

            if current_node.right_child:
                bfs_queue.put(current_node.right_child)

        return visited_nodes

bst = SearchableBinarySearchTree()

values = [50, 30, 70, 20, 40, 60, 80]

for value in values:
    bst.insert(value)

print("BFS:", bst.bfs())
```

```python
import queue

def bfs_traversal(graph, initial_vertex):
    visited = {initial_vertex}
    bfs_queue = queue.SimpleQueue()
    result = []

    bfs_queue.put(initial_vertex)

    while not bfs_queue.empty():
        current_vertex = bfs_queue.get()
        result.append(current_vertex)

        for adjacent_vertex in graph[current_vertex]:
            if adjacent_vertex not in visited:
                visited.add(adjacent_vertex)
                bfs_queue.put(adjacent_vertex)

    return result


print(bfs_traversal(graph, "A"))
```

### 9) Sort Algorithms

#### Bubble Sort

<details>
<summary>Compare adjacent elements and swap them until the list is sorted</summary>

**Idea:**
Compare neighboring elements and swap them if they are in the wrong order. After each pass, the largest remaining value “bubbles” to the end.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Learning/basic sorting logic
* Data is small
* Simplicity matters more than efficiency
* The list may already be nearly sorted

</details>

**Complexity:**
<details>

<summary>Show the answer</summary>

Optimized iterative version:

* Best: O(n) — already sorted
* Average: O(n²)
* Worst: O(n²)
* Space: O(1)

Basic version:

* Best/Average/Worst: O(n²)
* Space: O(1)

</details>

**Important:**
After each full pass, the largest remaining unsorted element is in its correct position, so the next pass can be shorter.

</details>

```python
def bubble_sort(values):
    n = len(values)

    for i in range(n - 1):
        for j in range(n - 1 - i):
            if values[j] > values[j + 1]:
                values[j], values[j + 1] = values[j + 1], values[j]

    return values

print(bubble_sort([5, 7, 9, 1, 4, 2]))
```

```python
def bubble_sort_optimized(values):
    n = len(values)
    is_sorted = False

    while not is_sorted:
        is_sorted = True

        for i in range(n - 1):
            if values[i] > values[i + 1]:
                values[i], values[i + 1] = values[i + 1], values[i]
                is_sorted = False

        # Last element of this pass is already in place
        n -= 1

    return values

print(bubble_sort_optimized([5, 7, 9, 1, 4, 2]))
```

#### Selection Sort

<details>
<summary>Find the smallest remaining element and move it to the front</summary>

**Idea:**
Repeatedly find the smallest element in the unsorted part of the list and swap it into the next correct position.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Learning/basic sorting logic
* Data is small
* You want an in-place algorithm with few swaps
* Simplicity matters more than speed

</details>

**Complexity:**
<details>

<summary>Show the answer</summary>

* Best: O(n²)
* Average: O(n²)
* Worst: O(n²)
* Space: O(1)

</details>

**Important:**
Selection Sort always scans the remaining unsorted part to find the minimum, even if the list is already sorted. After each pass, the next smallest element is placed in its final position.

</details>

```python
def selection_sort(my_list):
  list_length = len(my_list)

  for i in range(list_length - 1):
    # set lowest to the element of the list located at index i
    lowest = my_list[i]
    index = i

    # iterate again over the list starting on the next position of the i variable
    for j in range(i+1, list_length):
      # compare whether the element of the list located at index j is smaller than lowest
      if my_list[j] < lowest:
        index = j
        lowest = my_list[j]

    my_list[i] , my_list[index] = my_list[index] , my_list[i]
  
  return my_list

my_list = [6, 2, 9, 7, 4, 8] 
selection_sort(my_list)
print(my_list)
```

#### Insertion Sort

<details>
<summary>Take the next element and insert it into the correct place in the sorted part</summary>

**Idea:**
Build a sorted section from left to right. Take each new element, shift larger elements to the right, and insert it into the correct position.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Data is small
* Data is already or nearly sorted
* Elements arrive incrementally
* A simple in-place sort is enough

</details>

**Complexity:**
<details>

<summary>Show the answer</summary>

* Best: O(n) - Already sorted
* Average: O(n²)
* Worst: O(n²)
* Space: O(1)

</details>

**Important:**
Instead of repeatedly swapping, Insertion Sort saves the current value, shifts larger values right, and then inserts the saved value into the gap.

</details>

```python
def insertion_sort(values):
    for i in range(1, len(values)):
        value_to_insert = values[i]
        j = i - 1

        while j >= 0 and value_to_insert < values[j]:
            values[j + 1] = values[j]
            j -= 1

        values[j + 1] = value_to_insert

    return values

print(insertion_sort([5, 2, 9, 1, 4]))
```

#### Merge Sort

<details>
<summary>Split into halves, sort each half, then merge them back together</summary>

**Idea:**
Repeatedly divide the list into two halves until each part contains one element. Then merge the parts back together in sorted order.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Need predictable O(n log n) sorting
* Working with larger datasets
* A divide-and-conquer approach fits the problem
* Stable sorting is useful

</details>

**Complexity:**
<details>

<summary>Show the answer</summary>

* Best: O(n log n)
* Average: O(n log n)
* Worst: O(n log n)
* Space: O(n)

</details>

**Important:**
Merge Sort has two main steps:
1. Split recursively until only single elements remain.
2. Merge sorted halves by repeatedly taking the smaller front element.

Unlike Bubble, Selection, and Insertion Sort, it needs extra memory for the temporary halves.

</details>

```python
def merge_sort(values):
    if len(values) <= 1:
        return

    # split
    mid = len(values) // 2
    left_half = values[:mid]
    right_half = values[mid:]

    # sort both halves recursively
    merge_sort(left_half)
    merge_sort(right_half)

    # merge
    i = 0  # pointer for left_half
    j = 0  # pointer for right_half
    k = 0  # pointer for values

    while i < len(left_half) and j < len(right_half):
        if left_half[i] <= right_half[j]:
            values[k] = left_half[i]
            i += 1
        else:
            values[k] = right_half[j]
            j += 1

        k += 1

    # copy anything remaining on the left
    while i < len(left_half):
        values[k] = left_half[i]
        i += 1
        k += 1

    # copy anything remaining on the right
    while j < len(right_half):
        values[k] = right_half[j]
        j += 1
        k += 1

values = [35, 22, 90, 4, 50, 20, 30, 40, 1]
merge_sort(values)
print(values)
```

#### Quicksort

<details>
<summary>Choose a pivot, split smaller and larger values around it, then sort each side</summary>

**Idea:**
Choose a pivot and partition the list so smaller values go to one side and larger values go to the other. Then recursively apply the same process to both parts.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Need an efficient general-purpose sorting algorithm
* Average-case speed matters
* In-place sorting is useful
* Extra O(n) temporary arrays should be avoided

</details>

**Complexity:**
<details>

<summary>Show the answer</summary>

* Best: O(n log n)
* Average: O(n log n)
* Worst: O(n²)
* Extra space: O(log n) from recursion on average

</details>

**Important:**
Performance depends on the pivot. Bad pivots can create very uneven partitions and lead to O(n²) behavior.

</details>

```python
def partition(values, first, last):
    pivot = values[first]
    left = first + 1
    right = last

    while True:
        # move left pointer until finding a value >= pivot
        while left < last and values[left] < pivot:
            left += 1

        # move right pointer until finding a value <= pivot
        while values[right] > pivot:
            right -= 1

        # pointers crossed
        if left >= right:
            break

        # put values on the correct sides
        values[left], values[right] = values[right], values[left]

    # put pivot into its partition position
    values[first], values[right] = values[right], values[first]

    return right

def quicksort(values, first, last):
    if first < last:
        partition_index = partition(values, first, last)

        quicksort(values, first, partition_index)
        quicksort(values, partition_index + 1, last)

values = [35, 22, 90, 4, 50, 20, 30, 40, 1]
quicksort(values, 0, len(values) - 1)
print(values)
```
