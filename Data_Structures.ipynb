# Data Structures and Algorithms

### 1) Linked List

<details>

<summary>A chain of nodes: each node knows where the next one is</summary>

**Idea:**
A chain of **nodes**. Each node stores a value and a reference to the next node.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Frequent insertions/deletions
* Following `next` pointers
* Reversing a list
* Detecting a cycle
* Finding the middle

</details>

**Python:**
Python has no built-in linked list, so it is usually represented using a custom node:

`Node(value, next)`

**Complexity:**
<details>
<summary>Show the answer</summary>

* Access/Search: `O(n)`
* Insert/Delete when the location is known: `O(1)`
* If the location must first be found: `O(n)`

</details>

**Important:**
Unlike a Python list, there is no `O(1)` random access. To reach a node, start at the head and follow `next`.

</details>

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList():
    def __init__(self):
        self.head = None
        self.tail = None

    def insert_at_beginning(self, data):
        new_node = Node(data)
        if self.head:
            new_node.next = self.head
            self.head = new_node
        else:
            self.head = new_node
            self.tail = new_node

    def remove_at_beginning(self):
        self.head = self.head.next
```

**Q1) Middle Node:**&nbsp;&nbsp;Given the head of a singly linked list, return the middle node.

<details>
<summary>Show answer</summary>

Use two pointers:
- `slow` moves 1 step
- `fast` moves 2 steps
- When `fast` reaches the end, `slow` is at the middle

</details>

```python
def find_middle(head):
    slow = head
    fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

    return slow
```

### 2) Stack

<details>
<summary>LIFO (last in, first out)</summary>

**Idea:**
A singly linked list of elements with LIFO. Elements are added and removed from the top. The most recently added element is the first one removed (like a stack of books).

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Need the most recent item first
* Undo/history operations
* Matching or nested structures
* DFS / backtracking
* Reversing or processing things in reverse order

</details>

**Python:**
DataCamp uses Python's LifoQueue:

`queue.LifoQueue()`

A regular Python list can also be used as a stack with append() and pop().

**Complexity:**
<details>
<summary>Show the answer</summary>

* Access/Search: `O(n)`
* Push: `O(1)`
* Pop: `O(1)`
* Peek(top): `O(1)`

</details>

**Important:**
A stack only gives direct access to the top element. To reach an item deeper in the stack, elements above it must be processed first.

</details>

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class Stack:
    def __init__(self):
        self.top = None
        self.size = 0

    def push(self, data):
        new_node = Node(data)
        if self.top:
            new_node.next = self.top
        self.top = new_node
        self.size += 1   

    def pop(self):
        if self.top is None:
            return None
        else:
            popped_node = self.top
            self.size -=1
            self.top = self.top.next
            popped_node.next = None
            return popped_node.data

    def peek(self):
        if self.top:
            return self.top.data
        else:
            return None
```

```python
# use Python's built-in LifoQueue class to implement a stack
import queue

# create an infinite stack using LifoQueue
stack = queue.LifoQueue(maxsize=0)

# add elements to the stack
stack.put("first")
stack.put("second")

# get element
print("Top element:", stack.get())

# show the size
print("Size of stack:", stack.qsize())

# check if empty
print("Empty stack: ", stack.empty())
```

### 3) Queue

<details>
<summary>FIFO (first in, first out)</summary>

**Idea:**
Elements are added at the back and removed from the front. The first element added is the first one removed.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Items must be processed in arrival order
* Tasks/jobs waiting to be handled
* BFS / level-by-level traversal
* Scheduling or buffering

</details>

**Python:**
DataCamp uses Python's SimpleQueue:

`queue.SimpleQueue()`

A `collections.deque` is also commonly used as a queue.

**Complexity:**
<details>
<summary>Show the answer</summary>

* Enqueue: O(1)
* Dequeue: O(1)
* Access front: O(1)
* Search: O(n)

</details>

**Important:**
A queue follows FIFO: elements leave in the same order they entered.

</details>

```python
class Node():
    def __init__(self, data):
        self.data = data
        self.next = None

class Queue():
    def __init__():
        self.head = None
        self.tail = None
        
    def enqueue(self, data):
        new_node = Node(data)
        if self.head == None:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            self.tail = new_node 
    
    def dequeue(self):
        if self.head:
            current_node = self.head
            self.head = current_node.next
            current_node.next = None
        else:
            self.tail = None
```

```python
# use Python's built-in SimpleQueue class to implement a queue
import queue

# create an orders queue using SimpleQueue
orders_queue = queue.SimpleQueue()

# add elements to the queue
orders_queue.put("first")
orders_queue.put("second")
orders_queue.put("third")

# get element
print("Top element:", orders_queue.get())

# show the size
print("Size of queue:", orders_queue.qsize())

# check if empty
print("Empty queue: ", orders_queue.empty())
```

### 4) Hash Table

<details>
<summary>Store key → value pairs for fast lookup</summary>

**Idea:**
A hash table stores data using keys. A hash function maps each key to a location, allowing very fast lookup, insertion, and deletion on average.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Need fast lookup by key
* Checking whether something has been seen before
* Counting frequencies
* Mapping one value to another
* Finding complements or matches

</details>

**Python:**
Use Python's built-in dictionary: `dict`

Common operations:
`d[key]`
`d[key] = value`
`key in d`
`d.get(key)`
`d.pop(key)`

**Complexity:**
<details>
<summary>Show the answer</summary>

* Lookup: O(1) average
* Insert: O(1) average
* Delete: O(1) average
* Search by value: O(n)

</details>

**Important:**
The fast O(1) behavior applies to operations using keys. Looking for a value still usually requires checking the dictionary entries.

</details>

```python
# create a dictionary to store phone codes for different countries
phone_codes = {
    "France": "+33",
    "Germany": "+49",
    "Italy": "+39",
    "Spain": "+34",
    "Turkey": "+90",
}

# get items
print("Phone code for Germany:", phone_codes["Germany"])
print("Phone code for USA:", phone_codes.get("USA"))

print("Keys:", phone_codes.keys())
print("Values:", phone_codes.values())
print("Items:", phone_codes.items())

# delete an item
del phone_codes["France"]

# empty the dictionary
phone_codes.clear()

# delete the dictionary completely
del phone_codes
```

### 5) Tree

<details>
<summary>A hierarchical graph with no cycles</summary>

**Idea:**
A tree is a connected graph with no cycles. Nodes have parent-child relationships, starting from a root.
A binary tree allows each node to have at most two children.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Data has a hierarchical structure (like html)
* Parent-child relationships
* Searching/traversing levels or branches
* Problems involving paths, depth, or subtrees (file structure in computer)

</details>

**Python:**
Usually represented with a custom node class.

`Node(value, left, right)` for a binary tree.

**Complexity:**
<details>
<summary>Show the answer</summary>

* Traverse all nodes: O(n)
* Search in a general binary tree: O(n)
* Search in a balanced Binary Search Tree: O(log n) average

</details>

**Important:**
A binary tree is not automatically a Binary Search Tree (BST).
A BST additionally follows an ordering rule between the left child, node, and right child.

</details>

```python
class TreeNode:
    def __init__(self, data, left=None, right=None):
        self.data = data
        self.left_child = left
        self.right_child = right
    
    def is_leaf(self):
        return self.left_child is None and self.right_child is None

    def print_tree(self, level=0):
        if self.right_child:
            self.right_child.print_tree(level + 1)
        print(' ' * 4 * level + '->', self.data)
        if self.left_child:
            self.left_child.print_tree(level + 1)

node1 = TreeNode("B")
node2 = TreeNode("C")
root_node = TreeNode("A", node1, node2)
root_node.print_tree()
```

### 6) Graph

<details>
<summary>Nodes connected by edges</summary>

**Idea:**
A graph consists of nodes (vertices) connected by edges. Unlike trees, graphs may contain cycles and disconnected nodes. Graphs can be:
- Directed or undirected
- Weighted or unweighted
- Connected or disconnected

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Representing relationships or connections
* Networks, roads, social connections, dependencies
* Finding paths between nodes
* Exploring connected components
* Problems involving BFS or DFS

</details>

**Python:**
Often represented using an adjacency list, commonly with a dictionary:

`{node: [neighbors]}`

**Complexity:**
<details>
<summary>Show the answer</summary>

* BFS means Breadth-First Search. It explores a graph level by level, visiting all nearby nodes before moving farther away.
* DFS means Depth-First Search. It follows one path as far as possible before backtracking and trying another path.
* Both BFS and DFS usually visit every node and edge once, so their time complexity is O(V + E).
* Space means the extra memory needed while storing the graph and performing the traversal. With an adjacency list, the graph itself uses O(V + E) space.

V = number of vertices, or nodes
E = number of edges, or connection

</details>

**Important:**
A graph does not have to connect every node, and edges may have a direction or a weight depending on the problem.

</details>

```python
class Graph:
    def __init__(self):
        self.vertices = {}
    
    def add_vertex(self, vertex):
        if vertex not in self.vertices:
            self.vertices[vertex] = []
    
    def add_edge(self, source, target):
        if source in self.vertices and target in self.vertices:
            self.vertices[source].append(target)
            self.vertices[target].append(source)  # For undirected graph

    def add_directed_edge(self, source, target):
        if source in self.vertices and target in self.vertices:
            self.vertices[source].append(target)
    
    def add_weighted_edge(self, source, target, weight):
        if source in self.vertices and target in self.vertices:
            self.vertices[source].append((target, weight))
            self.vertices[target].append((source, weight))  # For undirected graph

my_graph = Graph()
my_graph.add_vertex('David')
my_graph.add_vertex('Miriam')
my_graph.add_vertex('Martin')
my_graph.add_edge('David', 'Miriam')
my_graph.add_edge('David', 'Martin')
my_graph.add_edge('Miriam', 'Martin')

print(my_graph.vertices)
```

### 7) Dynamic Programming

<details>
<summary>Save repeated work instead of calculating it again</summary>

**Idea:**
Break a problem into smaller overlapping subproblems and store their results so they are not recomputed.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* The same subproblems are solved repeatedly
* A recursive solution does lots of duplicate work
* A problem asks for an optimal value, count, or number of ways

</details>

**Common approaches:**

* Memoization (top-down): recursion + cache
* Tabulation (bottom-up): build results iteratively

**Important:**
DP often trades extra memory for much faster execution.

</details>

```python
cache = [None]*(100)

def fibonacci(n): 
    if n <= 1:
        return n
    
    # Check if the value exists
    if cache[n] is None:
        # Save the result in cache
        cache[n] = fibonacci(n-1) + fibonacci(n-2)
    return cache[n]

print(fibonacci(6))
```

```python
def hanoi(num_disks, from_rod, to_rod, aux_rod):
  # the base case
  if num_disks >= 1:
    # the calls to the hanoi function
    hanoi(num_disks - 1, from_rod, aux_rod, to_rod)
    print("Moving disk", num_disks, "from rod", from_rod,"to rod",to_rod)
    hanoi(num_disks - 1, aux_rod, to_rod, from_rod)   

num_disks = 4
source_rod = 'A'
auxiliar_rod = 'B'
target_rod = 'C'

hanoi(num_disks, source_rod, target_rod, auxiliar_rod)
```

### 8) Search Algorithms

#### Linear Search

<details>
<summary>Check elements one by one</summary>

**Idea:**
Go through the elements sequentially until the target is found.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Data is unsorted
* The collection is small
* No faster search structure is available

</details>

**Complexity:**
<details>

<summary>Show the answer</summary>

* Best case: O(1)
* Worst/Average case: O(n)

</details>

**Important:**
Works on both sorted and unsorted data, but may need to inspect every element.

</details>

```python
def linear_search(arr, target):
    for i, value in enumerate(arr):
        if value == target:
            return i
    return -1

print(linear_search([1,5,8,9,15,20,70,72], 5))
```

#### Binary Search

<details>
<summary>Repeatedly cut the search space in half</summary>

**Idea:**
Compare the target with the middle element and discard half of the remaining search space each step.

**Think of it when:**
<details>
<summary>Show the answer</summary>

* Data is sorted
* Fast searching is needed
* The search space can be repeatedly divided in half

</details>

**Complexity:**
<details>

<summary>Show the answer</summary>

* Best case: O(1)
* Worst/Average case: O(log n)

</details>

**Important:**
Binary search requires the data/search space to be ordered in a way that lets you safely eliminate half each step.

</details>

```python
def binary_search(ordered_list, target):
    first = 0
    last  = len(ordered_list) - 1
    
    while first <= last:
        middle = (first + last) // 2
        # check whether the search value equals the value in the middle
        if target == ordered_list[middle]:
            return True

        # check whether the search value is smaller than the value in the middle
        elif target < ordered_list[middle]:
            # set last to the value of middle minus one
            last = middle - 1

        else:
            first = middle + 1
    return False

print(binary_search([1,5,8,9,15,20,70,72], 5))
```

```python
def binary_search_recursive(ordered_list, target):
    # Define the base case
    if len(ordered_list) == 0:
        return False

    else:
        middle = len(ordered_list) // 2
        # check whether the search value equals the value in the middle
        if target == ordered_list[middle]:
            return True

        elif target < ordered_list[middle]:
            # call recursively with the left half of the list
            return binary_search_recursive(ordered_list[:middle], target)

        else:
            # call recursively with the right half of the list
            return binary_search_recursive(ordered_list[middle+1:], target)

print(binary_search_recursive([1,5,8,9,15,20,70,72], 5))
```
